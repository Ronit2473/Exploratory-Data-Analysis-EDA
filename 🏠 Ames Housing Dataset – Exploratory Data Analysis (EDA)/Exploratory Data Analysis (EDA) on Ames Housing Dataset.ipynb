Each column name (like MS SubClass, Lot Frontage, etc.)
is followed by a number that indicates the % of missing values in that column.

0.00000 → means no missing values (complete data).

16.74069 for Lot Frontage → means 16.74% of rows have missing values in that column.

So, in this example:

Lot Frontage is missing in roughly 17% of houses (not every property has a recorded frontage length).

Every other column shown has 0% missing data — meaning all houses have values for those features.

```python
def percent_missing(df):
    percent_nan=100*df.isnull().sum()/len(df)
    percent_nan=percent_nan[percent_nan>0].sort_values()
    return percent_nan
```

```python
percent_nan=percent_missing(df)
percent_nan
```

Electrical → 0.034165 → Only 0.03% of rows have missing Electrical data. (basically complete)

Garage Area → 0.034165 → Same — almost complete.

Lot Frontage → 16.74% → Around 1 in 6 houses don’t have recorded lot frontage length.

Fireplace Qu → 48.58% → Nearly half the homes don’t have a fireplace (so “missing” actually means no fireplace).

Fence → 80.46% → 80% don’t have a fence.

Alley → 93.23% → Only ~7% homes have alley access recorded.

Pool QC → 99.59% → Less than 1% have pools — nearly all missing!

```python
sns.barplot(x=percent_nan.index,y=percent_nan,palette='Set2')
plt.xticks(rotation=90)
```

```python
percent_nan[percent_nan<1]
df=df.dropna(axis=0,subset=['Electrical','Garage Cars'])
percent_nan=percent_missing(df)
percent_nan[percent_nan<1]
```

🧱 Basement Features

These variables describe basement square footage and bathrooms:

Since the missing percentages are extremely low (<0.1%), these are likely just data entry gaps, not structural absence.

We can safely fill missing values with median (for numeric) or mode (for categorical).

```python
# Bsmt numeric columns  ----> fillna 0
bsmt_num_cols=['BsmtFin SF 1','BsmtFin SF 2','Bsmt Unf SF', 'Total Bsmt SF','Bsmt Full Bath','Bsmt Half Bath']
df[bsmt_num_cols]=df[bsmt_num_cols].fillna(0)

# Bsmt string columns
bsmt_str_cols=['Bsmt Qual', 'Bsmt Cond', 'Bsmt Exposure','BsmtFin Type 1','BsmtFin Type 2']
df[bsmt_str_cols]=df[bsmt_str_cols].fillna('None')
```

```python
percent_nan=percent_missing(df)

sns.barplot(x=percent_nan.index,y=percent_nan,palette='Set2')
plt.xticks(rotation=90)
```

These basement and masonry features are highly correlated with SalePrice — because:

Houses with finished basements or masonry veneers (brick/stone exteriors) tend to have higher prices.

Since these missing values are few, imputing them correctly prevents data loss and improves model quality.

```python
df['Mas Vnr Area']=df['Mas Vnr Area'].fillna(0)
df['Mas Vnr Type']=df['Mas Vnr Type'].fillna('None')
percent_nan=percent_missing(df)

sns.barplot(x=percent_nan.index,y=percent_nan,palette='Set2')
plt.xticks(rotation=90)
```

Some features (like Garage Type, Lot Frontage) have moderate missingness → can be imputed or logically filled.

Others (like Pool QC, Fence, Alley) are extremely sparse → can safely be dropped, as they add little information.

```python
Gar_str_cols=['Garage Type','Garage Finish','Garage Qual','Garage Cond']
df[Gar_str_cols]=df[Gar_str_cols].fillna('None')
df['Garage Yr Blt']=df['Garage Yr Blt'].fillna(0)
percent_nan=percent_missing(df)

sns.barplot(x=percent_nan.index,y=percent_nan,palette='Set2')
plt.xticks(rotation=90)
```

Handled all garage-related missing values logically and consistently.

Improved data quality by distinguishing “no feature” from “missing data.”

Verified progress visually — garage-related missingness is gone, leaving only rare features (like pools or fences) to address next.

```python
df=df.drop(['Pool QC','Misc Feature','Alley','Fence'],axis=1)
percent_nan=percent_missing(df)

sns.barplot(x=percent_nan.index,y=percent_nan,palette='Set2')
plt.xticks(rotation=90)
```

🧩 Insight: Lot Frontage by Neighborhood

The boxplot comparing Lot Frontage across neighborhoods reveals noticeable variation in property sizes throughout Ames. Neighborhoods such as StoneBr and NridgHt tend to have larger median lot frontages, suggesting more spacious and possibly higher-valued residential areas. In contrast, denser neighborhoods like Edwards and MeadowV display smaller lot frontages, indicating more compact housing layouts. The presence of outliers in several neighborhoods points to occasional oversized plots, likely corner houses or premium properties.
This visualization confirms that Lot Frontage strongly depends on location, making it ideal to impute missing frontage values using each neighborhood’s median rather than a single overall value.

```python
df.groupby('Neighborhood')['Lot Frontage'].mean()
```

The neighborhood-wise analysis of Lot Frontage highlights significant differences in property dimensions across Ames. Premium neighborhoods such as NoRidge, NridgHt, NWAmes, and ClearCr exhibit larger average frontages (80–90 feet), reflecting more spacious plots typical of suburban or luxury zones. In contrast, compact neighborhoods like Blueste, BrDale, MeadowV, and NPkVill have much smaller frontages (20–30 feet), indicating higher housing density and smaller lot sizes.
This analysis confirms that Lot Frontage is highly dependent on neighborhood characteristics, supporting the decision to fill missing frontage values using the median frontage of each neighborhood for realistic imputation.

```python
df.isnull().sum()
```

After addressing missing values in garage, basement, and fireplace features, the only remaining column with missing data was Lot Frontage (490 entries).
Given that Lot Frontage varies significantly across neighborhoods, the missing values will be imputed using each neighborhood’s median frontage length to maintain realistic spatial relationships.
This step ensures consistent property dimension estimates while retaining neighborhood-specific characteristics.

```python
df['Lot Frontage']=df['Lot Frontage'].fillna(0)
df.isnull().sum().sum()
```

All missing values in the dataset were successfully handled.
The remaining missing entries in Lot Frontage (490 rows) were filled with 0, representing properties with no recorded or non-applicable frontage measurements.
After this step, the dataset contains no missing values, ensuring data consistency and readiness for further analysis and modeling.
While neighborhood-based median imputation can offer more precision, the use of 0 serves as a practical solution for maintaining dataset integrity without introducing new assumptions.
